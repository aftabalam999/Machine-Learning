```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
```

```python
dataset = pd.read_csv("email_spam_dataset.csv")
```

```python
dataset.isnull().sum()
# dataset.shape
dataset.info()
```

```python
dataset["output"] = dataset["output"].map({'ham': "not_spam", "spam": "spam"})
```

```python
dataset.head()
```

```python
inputData = dataset["input"]
outputData = dataset["output"]
```

```python
inputDataTrain, inputDataTest, outputDataTrain, outputDataTest = train_test_split(inputData, outputData, test_size=0.2, random_state=42)
```

```python
vectorizor = TfidfVectorizer()
conversion = vectorizor.fit_transform(inputDataTrain)
```

```python
# conversion.shape
print(conversion)
```

# Brain

```python
model = RandomForestClassifier()
```

# Model Train in conversion and outputDataTrain

```python
model.fit(conversion, outputDataTrain)
```

```python
new_msg = ["You have been chosen for an exclusive cash bonus of $1,000. RefS02399"]
```

```python
n_conversion = vectorizor.transform(new_msg)
# print(n_conversion)
```

```python
prediction = model.predict(n_conversion)
prediction
```
